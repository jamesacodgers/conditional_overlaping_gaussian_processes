```python
import pandas as pd
import numpy as np
import os 
from pathlib import Path
```

```python
# Case 2: Row-wise CV with entry masking on test set
seeds = [2026]

# Update model names and params for case 2 results
model_names_and_params = {
    'gplvm_ssvi_case2_cv_results': '_latentdim_6_inducing_50_maskseed_42_batchsize_32',
}

all_results = []
for model_name, params in model_names_and_params.items():
    for seed in seeds:
        file_path = Path(os.getcwd()) / 'results' / f"{model_name}_case2_seed_{seed}{params}.csv"
        try:
            df = pd.read_csv(file_path)
            df['model'] = model_name
            df['seed'] = seed
            all_results.append(df)
        except FileNotFoundError:
            print(f"File not found: {file_path}")
            continue
    
if all_results:
    all_results = pd.concat(all_results, ignore_index=True)
else:
    print("No results found. Make sure to run the case 2 experiments first.")
```

```python
# For each seed and model, get the mean RMSE and NLPD
# Note: In Case 2, 'test' refers to masked entries on test rows, 'train' refers to observed entries on test rows
if isinstance(all_results, pd.DataFrame) and len(all_results) > 0:
    summary_results = all_results.groupby(['model', 'seed']).agg({
        'total_rmse_test': 'mean',
        'total_nlpd_test': 'mean',
        'total_rmse_train': 'mean',
        'total_nlpd_train': 'mean',
        'n_train': 'first',
        'n_test': 'first'
    }).reset_index()
    
    summary_results_std = all_results.groupby(['model', 'seed']).agg({
        'total_rmse_test': 'std',
        'total_nlpd_test': 'std',
        'total_rmse_train': 'std',
        'total_nlpd_train': 'std'
    }).reset_index()
    
    print("Summary Results (Mean across folds):")
    display(summary_results)
else:
    print("No results to summarize.")
```

```python
# Overall summary across all seeds
if isinstance(all_results, pd.DataFrame) and len(all_results) > 0:
    overall_summary = all_results.groupby('model').agg({
        'total_rmse_test': 'mean',
        'total_nlpd_test': 'mean',
        'total_rmse_train': 'mean',
        'total_nlpd_train': 'mean'
    })
    print("Overall Summary (Mean across all seeds and folds):")
    display(overall_summary)
```

```python
# Per-dimension results
if isinstance(all_results, pd.DataFrame) and len(all_results) > 0:
    summary_results_incl_dims = all_results.groupby(['model', 'seed']).mean(numeric_only=True).reset_index()
    print("Per-dimension results (averaged across folds):")
    display(summary_results_incl_dims)
```
