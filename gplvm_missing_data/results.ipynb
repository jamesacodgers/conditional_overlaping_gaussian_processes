```python
import pandas as pd
import numpy as np
import os 
from pathlib import Path
```

```python
seeds = [1, 2, 3, 4, 5]

model_names_and_params = {'gplvm_cv_results':'_latentdim_5_inducing_50_maskseed_42', 
                          'gplvm_ssvi_case_1_cv_results':'_latentdim_6_inducing_50_maskseed_42_batchsize_32',
                          'knn_cv_results':'_neighbours_5_maskseed_42'
                          } 
all_results = []
for model_name, params in model_names_and_params.items():
    for seed in seeds:
        file_path = Path(os.getcwd()) / 'results' / f"{model_name}_seed_{seed}{params}.csv"
        try:
            df = pd.read_csv(file_path)
            df['model'] = model_name
            df['seed'] = seed
            all_results.append(df)
        except FileNotFoundError:
            print(f"File not found: {file_path}")
            continue
    
    
all_results = pd.concat(all_results, ignore_index=True)
```

```python
all_results
```

```python
# for each seed and model, get the mean rmse and nlpd and print in a nice dataframe
summary_results = all_results.groupby(['model', 'seed']).agg({'total_rmse_test':'mean', 'total_nlpd_test':'mean', 'total_rmse_train':'mean', 'total_nlpd_train':'mean'}).reset_index()
summary_results_std = all_results.groupby(['model', 'seed']).agg({'total_rmse_test':'std', 'total_nlpd_test':'std', 'total_rmse_train':'std', 'total_nlpd_train':'std'}).reset_index()


summary_results
```
